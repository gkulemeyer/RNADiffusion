```python
import pandas as pd
from pathlib import Path

base_path = Path("../data/ArchiveII_max_length_128.csv")
partition_path = Path("../data/famfold/ArchiveII_famfold.csv")

df = pd.read_csv(base_path, index_col="id")
partition_df = pd.read_csv(partition_path, index_col="id")
print("Base DataFrame:")
display(df.head(2))
print("\nPartition DataFrame:")
display(partition_df.head(2))
```

```python
df.query(" len < 128").fam.value_counts()
```

```python
train_ids = partition_df.loc[(partition_df["fold"] == "23s") & (partition_df["partition"] == "train")].index
common = list(set(train_ids).intersection(df.index))
df_train = df.loc[common]
print("\nTrain DataFrame:")
display(df_train.head(2))
df_train.shape
```

```python
train_ids = partition_df.loc[(partition_df["fold"] == "23s") & (partition_df["partition"] == "test")].index
common = list(set(train_ids).intersection(df.index))
df_train = df.loc[common]
print("\nTrain DataFrame:")
display(df_train.head(2))
df_train.shape
```
